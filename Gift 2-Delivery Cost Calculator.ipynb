TASK 15: To Create A Distance Charge Function

```python
def calculate_distance_charge(distance):
    if distance <= 10:
        return 500
    elif distance <= 20:
        return 1000
    elif distance <= 50:
        return 2000
    else:
        return 4000

calculate_distance_charge(15)
```

TASK 16: To Create A Weight Charge Function

```python
def calculate_weight_charge(package_weight):
    formatted_package_weight = (package_weight)
    if  formatted_package_weight  <= 2:
        weight_charge = 0
    elif  formatted_package_weight <= 5:
        weight_charge = 500
    elif  formatted_package_weight <= 50:
        weight_charge = 1000
    else:
        weight_charge = 2000
    return weight_charge

calculate_weight_charge(3)
    
```

TASK 17: To Create A Delivery Type Charge Function

```python
def calculate_delivery_type_charge(delivery_type):
    if delivery_type == "Standard":
        return 0
    elif delivery_type == "Express":
        return 1500
    else:
        return 3000

calculate_delivery_type_charge("Standard")
```

TASK 18: To Create Location Charge Function

```python
def calculate_location_charge(location_category):
    if location_category == "Local":
        return 0
    elif location_category == "Regional":
        return 1000
    else:
        return 2500

calculate_location_charge("Inter-State")
```

TASK 19: To Create A Discount Function

```python
def calculate_discount(sub_total, customer_type):
    if customer_type == "Premium":
        discount = sub_total * 0.10
    else:
        discount = 0
    return discount

calculate_discount(sub_total, customer_type)
```

TASK 20: To Create A Final Cost Function

```python
def calculate_final_cost(final_cost):
    final_cost = sub_total - discount
    return final_cost

calculate_final_cost(final_cost)
```
